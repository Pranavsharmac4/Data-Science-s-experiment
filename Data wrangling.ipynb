```python
import numpy as np
import pandas as pd

# ---- Dataset 1: students and their marks (used for merge and join) ----
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Asha", "Ravi", "Neha", "Imran", "Kiran", "Meera", "Arjun", "Sara"],
    "Department": ["CSE", "DS", "CSE", "DS", "IT", "IT", "CSE", "DS"]
})
marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105, 106, 109],
    "Data_Science_Marks": [86, 91, 78, 64, 72, 88]
})

# ---- Dataset 2: attendance and final marks of the same 8 students ----
attendance = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Attendance_Percent": [92, 78, 85, 96, 70, 88, 81, 90]
})
results = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Final_Marks": [88, 73, 81, 94, 60, 79, 76, 91]
})

# ---- Dataset 3: department-wise average marks ----
dept_marks = pd.DataFrame({
    "Department": ["CSE", "CSE", "CSE", "DS", "DS", "DS", "IT", "IT", "IT"],
    "Semester": [1, 2, 3, 1, 2, 3, 1, 2, 3],
    "Average_Marks": [76, 81, 79, 82, 85, 88, 70, 74, 72]
})
marks_by_sem = pd.DataFrame({
    "Semester_1": [76, 82, 70],
    "Semester_2": [81, 85, 74],
    "Semester_3": [79, 88, 72]
}, index=["CSE", "DS", "IT"])

# ---- Dataset 4: sales ----
jan = pd.DataFrame({"Month": ["Jan"] * 3, "Product": ["Pen", "Notebook", "Pencil"], "Units": [120, 80, 60]})
feb = pd.DataFrame({"Month": ["Feb"] * 3, "Product": ["Pen", "Notebook", "Pencil"], "Units": [150, 95, 70]})
mar = pd.DataFrame({"Month": ["Mar"] * 3, "Product": ["Pen", "Notebook", "Pencil"], "Units": [130, 110, 90]})
transactions = pd.DataFrame({
    "Product": ["Pen", "Pen", "Pen", "Notebook", "Notebook", "Notebook", "Pencil"],
    "Month": ["Jan", "Jan", "Feb", "Jan", "Feb", "Feb", "Jan"],
    "Units": [50, 70, 150, 80, 70, 25, 60]
})

# ---- Dataset 5: quiz scores (out of 10) ----
quiz_wide = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Neha"],
    "Quiz_1": [8, 7, 9],
    "Quiz_2": [9, 8, 10],
    "Quiz_3": [7, 9, 9]
})
```

```python
print("--- Students Head & Shape ---")
print(students.head())
print("Shape:", students.shape)

print("\n--- Marks Head & Shape ---")
print(marks.head())
print("Shape:", marks.shape)

print("\n--- Students Info ---")
students.info()

print("\n--- Missing Values ---")
print("Students missing:\n", students.isna().sum())
print("Marks missing:\n", marks.isna().sum())

print("\nIs Student_ID unique in students?:", students["Student_ID"].is_unique)
```

```python
multi = dept_marks.set_index(["Department", "Semester"])
print(multi)
print("\nIndex names:", multi.index.names)
```

```python
print("--- All CSE records ---")
print(multi.loc["CSE"])

print("\nCSE Semester 2 Average Marks:", multi.loc[("CSE", 2), "Average_Marks"])
print("DS Semester 3 Average Marks:", multi.loc[("DS", 3), "Average_Marks"])
```

```python
print("--- Reset both levels ---")
print(multi.reset_index())

print("\n--- Reset only Semester level ---")
print(multi.reset_index(level="Semester"))
```

```python
multi_sorted = multi.sort_index()
swapped = multi_sorted.swaplevel(0, 1)
swapped_sorted = swapped.sort_index()
print(swapped_sorted)
```

```python
print("--- Without ignore_index ---")
print(pd.concat([jan, feb]))

print("\n--- With ignore_index=True ---")
print(pd.concat([jan, feb], ignore_index=True))
```

```python
all_sales = pd.concat([jan, feb, mar], ignore_index=True)
print("--- Combined Sales ---")
print(all_sales)

totals = all_sales.groupby("Product")["Units"].sum()
print("\n--- Total Units per Product ---")
print(totals)
```

```python
att_idx = attendance.set_index("Student_ID")
res_idx = results.set_index("Student_ID")

side_by_side = pd.concat([att_idx, res_idx], axis=1)
print(side_by_side)
print("\nShape:", side_by_side.shape)
```

```python
inner_df = pd.merge(students, marks, on="Student_ID", how="inner")
print(inner_df)
print("\nNumber of rows:", len(inner_df))
```

```python
left_df = pd.merge(students, marks, on="Student_ID", how="left")
print(left_df)

print("\n--- Missing Values ---")
print(left_df.isna().sum())

missing_students = left_df[left_df["Data_Science_Marks"].isna()]["Name"]
print("\nStudents with missing marks:\n", missing_students.tolist())
```

```python
for join_type in ["inner", "left", "right", "outer"]:
    merged = pd.merge(students, marks, on="Student_ID", how=join_type)
    print(f"{join_type.capitalize()} join rows: {len(merged)}")

outer_df = pd.merge(students, marks, on="Student_ID", how="outer")
print("\n--- Outer Join ---")
print(outer_df)

unmatched_ids = outer_df[outer_df["Name"].isna() | outer_df["Data_Science_Marks"].isna()]["Student_ID"]
print("\nStudent_IDs not in both tables:", unmatched_ids.tolist())
```

```python
student_info = students.set_index("Student_ID")
student_marks = marks.set_index("Student_ID")

index_joined = student_info.join(student_marks, how="left")
print(index_joined)
```

```python
sales = pd.concat([jan, feb], ignore_index=True)
pivoted_sales = sales.pivot(index="Product", columns="Month", values="Units")
print(pivoted_sales)
```

```python
# Attempting pivot() raises a ValueError
try:
    transactions.pivot(index="Product", columns="Month", values="Units")
except ValueError as e:
    print("Error:", e)

# Fix using pivot_table
fixed_pivot = transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="sum")
print("\n--- Fixed with pivot_table ---")
print(fixed_pivot)
```

```python
pt_sum = transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="sum")
print("--- Sum with NaN ---")
print(pt_sum)

pt_zero = transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="sum", fill_value=0)
print("\n--- Sum with fill_value=0 ---")
print(pt_zero)

pt_mean = transactions.pivot_table(index="Product", columns="Month", values="Units", aggfunc="mean")
print("\n--- Average units per sale ---")
print(pt_mean)
```

```python
quiz_long = quiz_wide.melt(id_vars="Student", var_name="Quiz", value_name="Score")
print(quiz_long)
print("\nShape:", quiz_long.shape)
```

```python
quiz_long = quiz_wide.melt(id_vars="Student", var_name="Quiz", value_name="Score")

print("--- Average Score per Quiz ---")
print(quiz_long.groupby("Quiz")["Score"].mean())

print("\n--- Average Score per Student ---")
print(quiz_long.groupby("Student")["Score"].mean())
```

```python
unstacked_marks = multi["Average_Marks"].unstack()
print(unstacked_marks)
```

```python
stacked = marks_by_sem.stack()
print("--- Stacked (Series with MultiIndex) ---")
print(stacked)

unstacked_back = stacked.unstack()
print("\n--- Unstacked Back to Original ---")
print(unstacked_back)
```

```python
analysis_df = (
    students
    .merge(attendance, on="Student_ID", how="left")
    .merge(results, on="Student_ID", how="left")
)
print(analysis_df)
print("\nShape:", analysis_df.shape)
```

```python
dept_summary = analysis_df.groupby("Department")[["Attendance_Percent", "Final_Marks"]].mean().round(1)
print(dept_summary)
```

```python
dept_summary = analysis_df.groupby("Department")[["Attendance_Percent", "Final_Marks"]].mean().round(1)
print(dept_summary)
```

```python
bad_marks = pd.concat([marks, marks.iloc[[1]]], ignore_index=True)

bad_merge = pd.merge(students, bad_marks, on="Student_ID", how="left")
print(f"Row count with duplicate key: {len(bad_merge)} (Original was {len(students)})")

try:
    pd.merge(students, bad_marks, on="Student_ID", how="left", validate="one_to_one")
except pd.errors.MergeError as e:
    print("\nMergeError caught:", e)
```

```python
analysis_df["Attendance_Group"] = np.where(analysis_df["Attendance_Percent"] >= 85, "High", "Low")

att_pivot = analysis_df.pivot_table(
    index="Department",
    columns="Attendance_Group",
    values="Final_Marks",
    aggfunc="mean"
)
print(att_pivot)
```
